```python
import os
from google.colab import files
!pip install biopython

from Bio import SeqIO, Entrez
from Bio.SeqRecord import SeqRecord
Entrez.email = 'varshag.bt23@rvce.edu.in'
print("Searching NCBI for mRNA sequences related to 'hyperthyroid'...")
mysearch = Entrez.esearch(db='nucleotide', term="hyperthyroid AND mRNA[Filter]", retmax=5)
myrecord = Entrez.read(mysearch)
mysearch.close()
myids = myrecord["IdList"]
print(f"\nFound {len(myids)} mRNA records.\n")


```

```python
for id1 in myids:
    filename = id1 + "_mRNA.fasta"
    if not os.path.isfile(filename):
        print(f"Downloading {filename} ...")
        with Entrez.efetch(db='nucleotide', id=id1, rettype="fasta", retmode="text") as net_handle:
            with open(filename, "w") as out_handle:
                out_handle.write(net_handle.read())
        files.download(filename)
```

```python
print("\n--- mRNA Sequence Details ---\n")
for filename in os.listdir():
    if filename.endswith("_mRNA.fasta"):
        record = SeqIO.read(filename, "fasta")
        print(f"ID: {record.id}")
        print(f"Description: {record.description}")
        print(f"Sequence: {str(record.seq)} nucleotides\n")

```
